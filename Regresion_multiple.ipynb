## Bibliotecas y datos


```python
# Importar las bibliotecas matemáticas y de datos
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
# Procesamos los datos categóricos antes del entrenamiento
from sklearn.preprocessing import LabelEncoder, OneHotEncoder #
from sklearn.compose import ColumnTransformer #

# Importar el dataset
conjuntoDeDatos = pd.read_csv("80_Startups_Nuevo.csv") 

X = conjuntoDeDatos.iloc[:, :-1].values 

# Extraer el vector de la variable dependiente
y = conjuntoDeDatos.iloc[:, 4].values 
```

```python
# Aplicamos la codificacion 
ct = ColumnTransformer([('State', OneHotEncoder(categories='auto'), [3])], remainder='passthrough')
X = ct.fit_transform(X).astype(float)

# Eliminamos la primer columna que es una variable dummy (n-1 variables)
X = X[:, 1:]
```

```python
# Importamos sklearn
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=0)

# Ajustar el modelo con el conjunto de entrenamiento
regresion = LinearRegression()
regresion.fit(X_train, y_train)

# Prediccion
y_pred = regresion.predict(X_test)
```

```python

import statsmodels.api as sm


X = np.append(arr=np.ones((len(X), 1)).astype(int), values=X, axis=1)

# variable de significación
SL = 0.05

# vector de caracteristico
X_opt = X[:, [0, 1, 2, 3, 4, 5]].tolist()

# modelo de regresion
regresion_OLS = sm.OLS(endog=y, exog=X_opt).fit()


print(regresion_OLS.summary())
```

```python
# Entrenamiento del modelo con las variables optimas
X_definitivo = X[:, [1, 4]]

X_train_opt, X_test_opt, y_train, y_test = train_test_split(X_definitivo, y, test_size=0.2, random_state=0)

# Entrenamos el modelo definitivo
regresion_final = LinearRegression()
regresion_final.fit(X_train_opt, y_train)

# Generamos las predicciones finales 
y_pred_final = regresion_final.predict(X_test_opt)

# Comparamos el rendimiento y ver la mejora
from sklearn.metrics import r2_score
print(f"R2 del modelo óptimo: {r2_score(y_test, y_pred_final):.4f}")
```
